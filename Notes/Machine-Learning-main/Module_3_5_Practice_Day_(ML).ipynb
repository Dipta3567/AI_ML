## A2. Manual Min-Max on a tiny set

### Given scores = [20, 25, 30, 50], scale to [0, 1] by hand. Show each step.

### ✅ Solution:
**Formula:**

$$x' = \frac{x - \min}{\max - \min}$$

* **min** = 20
* **max** = 50
* **range** = 30

---

**Scaled values:**
* 20 $\rightarrow (20-20)/30 = 0$
* 25 $\rightarrow 5/30 = 0.167$
* 30 $\rightarrow 10/30 = 0.333$
* 50 $\rightarrow 30/30 = 1$



Final : [ 0, 0.167, 0.333, 1 ]

```python
import numpy as np

# Data
scores = np.array([20, 25, 30, 50])

# Manual Steps
score_min = np.min(scores)
score_max = np.max(scores)
score_range = score_max - score_min

# Transformation
scores_scaled = (scores - score_min) / score_range

print(f"Min: {score_min}, Max: {score_max}, Range: {score_range}")
print(f"Scaled Values: {np.round(scores_scaled, 3)}")
```

## A3. Z-scores on subset

### Given x = [8, 9, 11], compute mean, standard deviation, then standardize each. Use population
standard deviation for this question.

### ✅ Solution:

**Mean:**
$$\mu = \frac{8 + 9 + 11}{3} = 9.33$$

**Population std:**
$$\sigma = \sqrt{\frac{(8 - 9.33)^2 + (9 - 9.33)^2 + (11 - 9.33)^2}{3}} = 1.247$$

---

**Z-scores:**
* **8** $\rightarrow \frac{8 - 9.33}{1.247} = -1.069$
* **9** $\rightarrow \frac{9 - 9.33}{1.247} = -0.267$
* **11** $\rightarrow \frac{11 - 9.33}{1.247} = 1.336$

```python
import numpy as np

# Data
x = np.array([8, 9, 11])

# Calculate Mean and Population Std Dev (ddof=0)
mu = np.mean(x)
sigma = np.std(x, ddof=0)

# Calculate Z-scores
z_scores = (x - mu) / sigma

print(f"Mean (mu): {mu:.2f}")
print(f"Std Dev (sigma): {sigma:.3f}")
print(f"Z-scores: {np.round(z_scores, 3)}")
```

## A4. Robust scaling ingredients

### Given y = [5, 6, 6, 7, 50], find median, Q1, Q3, IQR. Do not scale yet.

### ✅ Solution:

**1. Sorted Data:**
$$[5, 6, 6, 7, 50]$$
The median ($Q_2$) is **6**.

---

**2. Calculate Quartiles:**

**Lower Half:** $[5, 6]$
$$Q_1 = \frac{5 + 6}{2} = 5.5$$

**Upper Half:** $[7, 50]$
$$Q_3 = \frac{7 + 50}{2} = 28.5$$

---

**3. Calculate IQR:**
$$\text{IQR} = Q_3 - Q_1$$
$$\text{IQR} = 28.5 - 5.5 = 23$$

---

**Scaled values:**
*(If you are applying the formula from earlier to this set, you can list them here)*

```python
import numpy as np

# Data
y = [5, 6, 6, 7, 50]

# Manual calculation of quartiles for small sets
def get_quartiles(data):
    sorted_data = sorted(data)
    n = len(sorted_data)
    mid = n // 2

    median = sorted_data[mid]
    # Q1 is median of the lower half
    lower_half = sorted_data[:mid]
    q1 = np.median(lower_half)

    # Q3 is median of the upper half
    upper_half = sorted_data[mid+1:]
    q3 = np.median(upper_half)

    return q1, median, q3

q1, median, q3 = get_quartiles(y)
iqr = q3 - q1

print(f"Sorted Data: {sorted(y)}")
print(f"Q1: {q1}, Median: {median}, Q3: {q3}, IQR: {iqr}")
```

### ✅ Task B3: Ordinal Mapping

**Original Data:** Education = [High School, Bachelor, Master, Bachelor, Master]

---

**1. Initial Mapping ($0$ to $2$):**
* **High School** $\rightarrow 0$
* **Bachelor** $\rightarrow 1$
* **Master** $\rightarrow 2$

**Resulting Vector:** $[0, 1, 2, 1, 2]$

---

**2. Shifted Mapping ($1$ to $3$):**
* **High School** $\rightarrow 1$
* **Bachelor** $\rightarrow 2$
* **Master** $\rightarrow 3$

**Resulting Vector:** $[1, 2, 3, 2, 3]$

---

**3. Impact of the Shift:**
**Analysis:** Shifting the mapping from $(0, 1, 2)$ to $(1, 2, 3)$ changes the absolute values but **does not shift the relative distances** between the categories, as the difference ($d$) between levels remains constant ($1$ unit).

```python
import pandas as pd

# Original Data
education = ['High School', 'Bachelor', 'Master', 'Bachelor', 'Master']

# 1. Define Mapping A (0-2)
mapping_a = {'High School': 0, 'Bachelor': 1, 'Master': 2}

# 2. Define Mapping B (1-3)
mapping_b = {'High School': 1, 'Bachelor': 2, 'Master': 3}

# Apply mappings
mapped_a = [mapping_a[level] for level in education]
mapped_b = [mapping_b[level] for level in education]

# Display results
print(f"Original: {education}")
print(f"Mapping A (0-2): {mapped_a}")
print(f"Mapping B (1-3): {mapped_b}")

# Explanation of distances
diff_a = mapping_a['Bachelor'] - mapping_a['High School']
diff_b = mapping_b['Bachelor'] - mapping_b['High School']
print(f"\nDistance between HS and Bachelor in A: {diff_a}")
print(f"Distance between HS and Bachelor in B: {diff_b}")
print("Result: Shifting the map does not change the relative distance between levels.")
```

## B4. Encoding mixup

### You mistakenly apply ordinal encoding to City and one-hot to Education. Write one sentence on the risk this creates in a linear model.

### ⚠️ Task B4: The Risk of Encoding Mixup

**The Risk (One-Sentence Summary):**
> Applying ordinal encoding to **City** falsely implies a mathematical ranking where one city is "greater than" another (e.g., $\text{Rajshahi} > \text{Dhaka}$), while one-hot encoding **Education** hides the natural progression and relative distance between academic levels, causing a linear model to lose the inherent "order" of the data.

---

### 📉 Why this breaks a Linear Model:

#### **1. For City (Nominal Data):**
A linear model calculates a weight ($w$) for each feature. If Cities are $0, 1, 2$, the model assumes:
$$\hat{y} = w \cdot \text{City} + b$$
* This implies that **Rajshahi (2)** is "twice as much City" as **Chattogram (1)**.
* It forces a linear relationship ($0 \rightarrow 1 \rightarrow 2$) on geographical categories that have no numerical order, leading to nonsensical predictions.

#### **2. For Education (Ordinal Data):**
A linear model thrives on the "trend" of ordinal data (e.g., more education usually correlates with higher salary). By using One-Hot Encoding:
$$\text{Education} \rightarrow [x_1, x_2, x_3]$$
* The model treats **High School**, **Bachelor**, and **Master** as completely independent variables (like colors).
* It loses the mathematical information that:
$$\text{High School} < \text{Bachelor} < \text{Master}$$
* This forces the model to learn three separate weights instead of one cohesive trend, often requiring more data to reach the same accuracy.

```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder

# Original Data
data = pd.DataFrame({
    'City': ['Dhaka', 'Chattogram', 'Rajshahi'],
    'Education': ['High School', 'Bachelor', 'Master']
})

# MISTAKE 1: Ordinal Encoding on City (Randomly assigning numbers)
city_map = {'Dhaka': 0, 'Chattogram': 1, 'Rajshahi': 2}
data['City_Encoded'] = data['City'].map(city_map)

# MISTAKE 2: One-Hot Encoding on Education
encoder = OneHotEncoder(sparse_output=False)
edu_encoded = encoder.fit_transform(data[['Education']])
edu_df = pd.DataFrame(edu_encoded, columns=encoder.get_feature_names_out(['Education']))

# Result
print("--- The Encoding Mixup ---")
print(pd.concat([data[['City_Encoded']], edu_df], axis=1))
```

## B5. Vectors and alignment

### ✅ Task B5: Vectors and Alignment

**Given Vectors:**
$$\mathbf{a} = [3, -1, 2], \quad \mathbf{b} = [4, 0, -2], \quad \mathbf{c} = [-6, 2, -4]$$

---

#### **a) Compute Dot Products**
**Formula:** $\mathbf{a} \cdot \mathbf{b} = \sum a_i b_i$

* **$\mathbf{a} \cdot \mathbf{b}$:**
$$(3 \times 4) + (-1 \times 0) + (2 \times -2) = 12 + 0 - 4 = \mathbf{8}$$

* **$\mathbf{a} \cdot \mathbf{c}$:**
$$(3 \times -6) + (-1 \times 2) + (2 \times -4) = -18 - 2 - 8 = \mathbf{-28}$$

---

#### **b) Compare Signs and Alignment**
* **Alignment with $\mathbf{b}$:** Since $\mathbf{a} \cdot \mathbf{b} = 8 > 0$, the vectors are **generally aligned** (the angle between them is less than $90^\circ$).
* **Alignment with $\mathbf{c}$:** Since $\mathbf{a} \cdot \mathbf{c} = -28 < 0$, the vectors are **generally opposite** (the angle between them is greater than $90^\circ$).
* **Pro-tip:** Notice that $\mathbf{c} = -2\mathbf{a}$. This means they are perfectly **anti-parallel** (facing exactly opposite directions).

---

#### **c) L2 Normalization of $\mathbf{a}$**
**Formula:** $\hat{\mathbf{a}} = \frac{\mathbf{a}}{\|\mathbf{a}\|_2}$, where $\|\mathbf{a}\|_2 = \sqrt{3^2 + (-1)^2 + 2^2}$

1. **Calculate Magnitude:**
$$\|\mathbf{a}\|_2 = \sqrt{9 + 1 + 4} = \sqrt{14} \approx 3.742$$

2. **Normalize Vector:**
$$\hat{\mathbf{a}} = \left[ \frac{3}{3.742}, \frac{-1}{3.742}, \frac{2}{3.742} \right] \approx \mathbf{[0.802, -0.267, 0.535]}$$

```python
import numpy as np

# Define the vectors
a = np.array([3, -1, 2])
b = np.array([4, 0, -2])
c = np.array([-6, 2, -4])

# --- Task a: Dot Products ---
dot_ab = np.dot(a, b)
dot_ac = np.dot(a, c)

# --- Task b: Analyze Alignment ---
# Function to describe alignment
def check_alignment(dot_product, name):
    if dot_product > 0:
        return f"Positive ({dot_product}): a and {name} are generally aligned."
    elif dot_product < 0:
        return f"Negative ({dot_product}): a and {name} are generally opposite."
    else:
        return f"Zero: a and {name} are orthogonal (90 degrees)."

# --- Task c: L2 Normalization ---
magnitude_a = np.linalg.norm(a)
a_normalized = a / magnitude_a

# --- Print Final Results ---
print(f"Task A - Dot Product (a.b): {dot_ab}")
print(f"Task A - Dot Product (a.c): {dot_ac}")
print("-" * 30)
print(f"Task B - Alignment (b): {check_alignment(dot_ab, 'b')}")
print(f"Task B - Alignment (c): {check_alignment(dot_ac, 'c')}")
print("-" * 30)
print(f"Task C - Magnitude of a: {magnitude_a:.3f}")
print(f"Task C - Normalized Vector a: {np.round(a_normalized, 3).tolist()}")
```

## B6. Two distances, different vibes

### ✅ Task B6: Euclidean vs. Manhattan Distance

**Points:** $P_1(2, 3), \quad P_2(5, 7), \quad P_3(2, 10)$

---

#### **a) Distance Calculations for All Pairs**

| Pair | Manhattan ($|x_1-x_2| + |y_1-y_2|$) | Euclidean ($\sqrt{\Delta x^2 + \Delta y^2}$) |
| :--- | :--- | :--- |
| **$P_1, P_2$** | $|2-5| + |3-7| = 3 + 4 = \mathbf{7}$ | $\sqrt{3^2 + 4^2} = \sqrt{25} = \mathbf{5}$ |
| **$P_2, P_3$** | $|5-2| + |7-10| = 3 + 3 = \mathbf{6}$ | $\sqrt{3^2 + 3^2} = \sqrt{18} \approx \mathbf{4.243}$ |
| **$P_1, P_3$** | $|2-2| + |3-10| = 0 + 7 = \mathbf{7}$ | $\sqrt{0^2 + 7^2} = \sqrt{49} = \mathbf{7}$ |

---

#### **b) Sensitivity Analysis**
**Question:** Which is more sensitive to a single large jump in one coordinate?  
**Answer:** **Manhattan distance** is more sensitive to a single large jump. Because Euclidean distance squares the differences before taking the root, it effectively "shortcuts" the diagonal, whereas Manhattan forces you to travel the full length of that large jump.

---

#### **c) Scaling Effect ($y \times 10$)**
**New Points:** $P_1(2, 30), \quad P_2(5, 70)$  
* **New Manhattan:** $|2-5| + |30-70| = 3 + 40 = \mathbf{43}$
* **New Euclidean:** $\sqrt{3^2 + 40^2} = \sqrt{1609} \approx \mathbf{40.112}$

**One-line Explanation:** Scaling a single dimension ($y$) by 10 causes the distance metric to be dominated by that dimension, effectively making the $x$-coordinate's contribution negligible.

```python
import numpy as np

# Define Points
p1 = np.array([2, 3])
p2 = np.array([5, 7])
p3 = np.array([2, 10])

def get_distances(a, b):
    manhattan = np.sum(np.abs(a - b))
    euclidean = np.linalg.norm(a - b)
    return manhattan, euclidean

# Task a: Pairs
pairs = [("P1-P2", p1, p2), ("P2-P3", p2, p3), ("P1-P3", p1, p3)]

print("--- Task A: Pairwise Distances ---")
for name, a, b in pairs:
    m, e = get_distances(a, b)
    print(f"{name}: Manhattan = {m}, Euclidean = {e:.3f}")

# Task c: Scaling y by 10
p1_scaled = np.array([2, 30])
p2_scaled = np.array([5, 70])

m_s, e_s = get_distances(p1_scaled, p2_scaled)

print("\n--- Task C: Scaled Distances (y * 10) ---")
print(f"P1-P2 Scaled: Manhattan = {m_s}, Euclidean = {e_s:.3f}")
print("Effect: The distance is now almost entirely determined by the Y-axis difference.")
```

## C3. Outlier stress test

Using Income_BDT in C-Data-2, compute Min-Max scaled values. Then compute Robust scaled
values. In one line, compare how each treats the 300000 outlier.

#### **A) Min-Max Scaling ($Income\_BDT$)**
**Formula:** $x' = \frac{x - \min}{\max - \min}$
* **Min:** $30,000$ | **Max:** $300,000$ | **Range:** $270,000$
* $30,000 \rightarrow 0.0$
* $45,000 \rightarrow \frac{45,000 - 30,000}{270,000} = \mathbf{0.056}$
* $300,000 \rightarrow \frac{300,000 - 30,000}{270,000} = \mathbf{1.0}$

#### **B) Robust Scaling ($Income\_BDT$)**
**Formula:** $x' = \frac{x - \text{Median}}{\text{IQR}}$
* **Median ($Q_2$):** $45,000$ | **$Q_1$:** $38,000$ | **$Q_3$:** $52,000$ | **IQR:** $14,000$
* $30,000 \rightarrow \frac{30,000 - 45,000}{14,000} = \mathbf{-1.071}$
* $300,000 \rightarrow \frac{300,000 - 45,000}{14,000} = \mathbf{18.214}$

#### **C) One-Line Comparison**
**Analysis:** Min-Max scaling squashes all normal incomes into a negligible range ($0$ to $0.08$) to fit the outlier, whereas Robust Scaling preserves the relative spread of the majority and marks the $300,000$ value as a clear extreme ($18$ units away).

```python
import numpy as np

# Data
income = np.array([30000, 45000, 52000, 300000, 38000])

# --- Task A: Min-Max ---
mm_scaled = (income - np.min(income)) / (np.max(income) - np.min(income))

# --- Task B: Robust ---
# Based on sorted list: [30k, 38k, 45k, 52k, 300k]
median = 45000
q1, q3 = 38000, 52000
iqr = q3 - q1
robust_scaled = (income - median) / iqr

# --- Task C: Results ---
print(f"Min-Max Scaled: {np.round(mm_scaled, 3)}")
print(f"Robust Scaled:  {np.round(robust_scaled, 3)}")
print("\nComparison: Min-Max is heavily biased by the 300k outlier, while Robust resists it.")
```

## C4. Distance on feature space

From C-Data-1, take feature pair (Hours_Study, GPA).

a) Compute the Euclidean distance between ID 1 and ID 4

b) Compute the Manhattan distance for the same pair

c) Normalize Hours_Study and GPA with Min-Max and recompute both distances, then comment in one line on scale effects

#### **A) Euclidean vs. Manhattan (Raw Data)**
**Points (Hours_Study, GPA):** $ID_1 = (1.0, 3.1), \quad ID_4 = (5.0, 3.9)$

* **Euclidean Distance ($d_E$):**
    $$\sqrt{(5.0 - 1.0)^2 + (3.9 - 3.1)^2} = \sqrt{4^2 + 0.8^2} = \sqrt{16.64} \approx \mathbf{4.079}$$
* **Manhattan Distance ($d_M$):**
    $$|5.0 - 1.0| + |3.9 - 3.1| = 4 + 0.8 = \mathbf{4.8}$$

#### **B) Normalized Distances (Min-Max)**
Using the full range of C-Data-1: `Hours_Study` ($0.2$ to $5.0$), `GPA` ($2.3$ to $3.9$).
* **Normalized $ID_1$:** $(\frac{1.0-0.2}{4.8}, \frac{3.1-2.3}{1.6}) \rightarrow \mathbf{(0.167, 0.5)}$
* **Normalized $ID_4$:** $(\frac{5.0-0.2}{4.8}, \frac{3.9-2.3}{1.6}) \rightarrow \mathbf{(1.0, 1.0)}$

**Recomputed Distances:**
* **New Euclidean:** $\sqrt{(1.0 - 0.167)^2 + (1.0 - 0.5)^2} = \sqrt{0.833^2 + 0.5^2} \approx \mathbf{0.971}$
* **New Manhattan:** $|1.0 - 0.167| + |1.0 - 0.5| = 0.833 + 0.5 = \mathbf{1.333}$

#### **C) One-Line Comment**
**Effect:** Normalizing the features prevents `Hours_Study` (which has a larger raw range) from completely dominating the distance calculation, giving `GPA` a fairer influence on the result.

```python
import numpy as np
from scipy.spatial import distance

# Original Points from C-Data-1
p1 = np.array([1.0, 3.1])
p4 = np.array([5.0, 3.9])

# --- Task A & B: Raw Distances ---
euc_raw = distance.euclidean(p1, p4)
man_raw = distance.cityblock(p1, p4)

# --- Task C: Normalized Distances ---
# Full ranges from C-Data-1
h_min, h_max = 0.2, 5.0
g_min, g_max = 2.3, 3.9

p1_norm = np.array([(p1[0]-h_min)/(h_max-h_min), (p1[1]-g_min)/(g_max-g_min)])
p4_norm = np.array([(p4[0]-h_min)/(h_max-h_min), (p4[1]-g_min)/(g_max-g_min)])

euc_norm = distance.euclidean(p1_norm, p4_norm)
man_norm = distance.cityblock(p1_norm, p4_norm)

# Results
print(f"Raw: Euclidean = {euc_raw:.3f}, Manhattan = {man_raw:.3f}")
print(f"Normalized: Euclidean = {euc_norm:.3f}, Manhattan = {man_norm:.3f}")
print("\nEffect: Normalization scales both features to [0,1], balancing their contribution.")
```
